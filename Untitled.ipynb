```python
from stretch.perception.encoders.mobile_clip_encoder import MaskMobileClipEncoder

m = MaskMobileClipEncoder(device = "cpu", version = "S2")
```

```python
from PIL import Image
import numpy as np
import torch

img = Image.open('tennis ball.jpg')
i = torch.from_numpy(np.asarray(img))
img
```

```python
image, features = m.run_mask_siglip(i.permute(2, 0, 1), (480, 360))
```

```python
from matplotlib import pyplot as plt

t_features = m.encode_text("purple computer fan")

sim = (features @ t_features[0].cpu())[0]

plt.imshow((sim - sim.min()) / (sim.max() - sim.min()))
plt.show()

plt.imshow(sim > 0.2)
plt.show()

sim.max()
```
